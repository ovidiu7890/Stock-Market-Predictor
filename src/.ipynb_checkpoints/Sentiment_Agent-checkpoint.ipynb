# The Sentiment Agent (The "Newsroom")
This notebook implements an SVM paired with TfidfVectorizer to process financial news and determine if the prevailing mood is *Bullish* or *Bearish*.

It converts raw language into a sentiment probability score.

```
import pandas as pd
import numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, accuracy_score
```

```
# Load the dataset
# Update the path if necessary. We are using the perfectly_balanced_stock_news.csv file based on your description.
df = pd.read_csv('csv-history/perfectly_balanced_stock_news.csv')
df.head()
```

```
# Preprocessing
# We will use the 'Article' column for text analysis.
# If your dataset doesn't have a labeled 'Sentiment' column (e.g., Bullish/Bearish), 
# you'll need to define how to create one (for example, based on stock price movement).
# For this notebook, let's assume there is a 'Sentiment' column where 1 = Bullish and 0 = Bearish.
# If it doesn't exist, you can generate proxy labels or load a labeled dataset.

# Drop missing values in the text column
df = df.dropna(subset=['Article'])

# Dummy example: Assuming you have a 'Sentiment' column. 
# If not, you must create it before training.
# Example: df['Sentiment'] = ... 

# Let's verify our text and target variables:
# X = df['Article']
# y = df['Sentiment']
```

```
# To make this notebook run out-of-the-box as a template, 
# we will create a mock 'Sentiment' column (Randomly assigned for demonstration).
# REPLACE THIS WITH YOUR ACTUAL SENTIMENT LABELS OR STOCK PRICE LOGIC.
np.random.seed(42)
df['Sentiment'] = np.random.choice(['Bullish', 'Bearish'], size=len(df))

X = df['Article']
y = df['Sentiment']

# Split the dataset into training and testing sets
# We use a small subset for demonstration purposes, remove `[:1000]` to train on the full dataset
X_train, X_test, y_train, y_test = train_test_split(X[:1000], y[:1000], test_size=0.2, random_state=42)
```

```
# TF-IDF Vectorization
# Convert raw text into numerical features
vectorizer = TfidfVectorizer(max_features=5000, stop_words='english')

X_train_tfidf = vectorizer.fit_transform(X_train)
X_test_tfidf = vectorizer.transform(X_test)
```

```
# Define and train the SVM Model
# We set probability=True so it can convert language into a sentiment probability score
svm_model = SVC(kernel='linear', probability=True, random_state=42)

print("Training the SVM model... This might take some time.")
svm_model.fit(X_train_tfidf, y_train)
print("Training complete!")
```

```
# Evaluation
y_pred = svm_model.predict(X_test_tfidf)

print("Accuracy Score:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```

```
# Convert raw language into a sentiment probability score
sample_texts = [
    "The company reported record profits and outstanding growth in Q3.",
    "Inflation concerns and supply chain disruptions have led to a massive drop in revenue."
]

sample_tfidf = vectorizer.transform(sample_texts)
predictions = svm_model.predict(sample_tfidf)
probabilities = svm_model.predict_proba(sample_tfidf)

for text, pred, prob in zip(sample_texts, predictions, probabilities):
    print(f"Text: {text}")
    print(f"Predicted Mood: {pred}")
    print(f"Sentiment Probabilities: {svm_model.classes_[0]}: {prob[0]:.4f}, {svm_model.classes_[1]}: {prob[1]:.4f}\n")
```
