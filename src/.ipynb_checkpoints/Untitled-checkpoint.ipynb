```python
from datasets import load_dataset
from tqdm.notebook import tqdm
import pandas as pd

target_companies = ["AAPL", "MSFT", "NVDA", "GOOGL", "AMZN", "WMT", "TSLA", "JPM", "BRKL", "LLY"]
TARGET_PER_COMPANY = 500  # 500 each * 10 companies = 5,000 total
MAX_TOTAL_GOAL = TARGET_PER_COMPANY * len(target_companies)

# 1. Load the dataset in streaming mode 
dataset = load_dataset(
    "csv", 
    data_files="https://huggingface.co/datasets/Zihan1004/FNSPID/resolve/main/Stock_news/nasdaq_exteral_data.csv", 
    split="train", 
    streaming=True
)

# 2. Initialize an empty list for EACH company to track them individually
news_by_company = {ticker: [] for ticker in target_companies}
row_counter = 0  

print("Starting stream...")

with tqdm(desc="Scanning rows", unit=" rows") as pbar:
    for row in dataset:
        row_counter += 1
        ticker = row.get('Stock_symbol')
        
        # Check if the ticker is in our target list AND hasn't hit its individual quota yet
        if ticker in news_by_company and len(news_by_company[ticker]) < TARGET_PER_COMPANY:
            news_by_company[ticker].append(row)
            
        # UI Update every 500 scanned rows
        if row_counter % 500 == 0:
            pbar.update(500)
            total_collected = sum(len(rows) for rows in news_by_company.values())
            pct_complete = (total_collected / MAX_TOTAL_GOAL) * 100
            pbar.set_postfix_str(f"Total Matches: {total_collected}/{MAX_TOTAL_GOAL} ({pct_complete:.1f}%)")
            
        # STOPPING CONDITION: Check if EVERY single company has hit its target
        if all(len(news_by_company[t]) >= TARGET_PER_COMPANY for t in target_companies): 
            pbar.update(row_counter % 500)
            print(f"\nSuccess! Collected exactly {TARGET_PER_COMPANY} rows for all companies.")
            break
            
    else:
        # Fail-safe: Executes only if the dataset naturally runs out of rows 
        # before every single company hits 500.
        pbar.update(row_counter % 500)
        print("\nReached the end of the dataset. Stopping stream.")

# 3. Combine all separate company lists back into one single flat list
filtered_news = []
for ticker, rows in news_by_company.items():
    filtered_news.extend(rows)

print(f"Done! You now have a balanced pool of {len(filtered_news)} rows in memory.")

# 4. Verify the distribution breakdown
print("\nFinal Data Breakdown:")
for ticker, rows in news_by_company.items():
    print(f" - {ticker}: {len(rows)} articles")
```

```python
import pandas as pd

# 1. Turn your data into a dataframe
df = pd.DataFrame(filtered_news)

# 2. Use '../' to step out of 'src' and look into 'csv-history'
df.to_csv("../csv-history/perfectly_balanced_stock_news.csv", index=False)

print("Saved successfully inside the csv-history folder!")
```
